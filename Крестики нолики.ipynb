```python
# Инициализация поля 3x3 (список списков)
field = [['-' for _ in range(3)] for _ in range(3)]

# Функция для вывода поля в консоль с номерами строк и столбцов
def print_field():
    # Выводим номера столбцов (0, 1, 2) в верхней строке
    print("  " + " ".join(str(i) for i in range(3)))

    # Проходим по строкам поля
    for i, row in enumerate(field):
        # Выводим номер строки слева и саму строку
        print(str(i) + " " + " ".join(row))

# Функция для проверки корректности хода
def is_valid_move(row, col):
    return 0 <= row < 3 and 0 <= col < 3 and field[row][col] == '-'

# Функция для проверки победы
def check_winner():
    # Проверяем строки, столбцы и диагонали
    lines = field + [[field[i][j] for i in range(3)] for j in range(3)] + \
            [[field[i][i] for i in range(3)], [field[i][2-i] for i in range(3)]]
    for line in lines:
        if line == ['X', 'X', 'X']:
            return 'X'
        elif line == ['O', 'O', 'O']:
            return 'O'
    return None

# Основная игра
current_player = 'X'  # Начинаем с крестика
game_over = False

while not game_over:
    # Убрали очистку консоли ("\n" * 100) — это создавало лишнее пространство
    print("\n")  # Оставляем только один перенос строки между ходами

    # Выводим поле с подписями строк и столбцов
    print_field()
    print(f"Ход игрока {current_player}")

    # Ввод координат (строка и столбец)
    try:
        row = int(input("Введите номер строки (0-2): "))
        col = int(input("Введите номер столбца (0-2): "))
    except ValueError:
        print("Ошибка: введите числа от 0 до 2!")
        continue

    # Проверка корректности хода
    if not is_valid_move(row, col):
        print("Неверный ход! Ячейка занята или координаты вне поля.")
        continue

    # Делаем ход
    field[row][col] = current_player

    # Проверяем победу
    winner = check_winner()
    if winner:
        print("\n")  # Перенос строки перед объявлением победителя
        print_field()
        print(f"Победил игрок {winner}!")
        game_over = True
    elif all(field[i][j] != '-' for i in range(3) for j in range(3)):
        print("\n")  # Перенос строки перед объявлением ничьей
        print_field()
        print("Ничья!")
        game_over = True
    else:
        # Переключаем игрока
        current_player = 'O' if current_player == 'X' else 'X'
```

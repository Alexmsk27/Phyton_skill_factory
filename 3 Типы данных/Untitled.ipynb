```python
import random

# список учеников
students = ['Аполлон', 'Ярослав', 'Александра', 'Дарья', 'Ангелина']
students.sort()

# список предметов
classes = ['Математика', 'Русский язык', 'Информатика']

# словарь с оценками
students_marks = {}

# генерация случайных оценок
for student in students:
    students_marks[student] = {}
    for class_ in classes:
        marks = [random.randint(1, 5) for _ in range(3)]
        students_marks[student][class_] = marks

# стартовый вывод
for student in students:
    print(f"{student}: {students_marks[student]}")

print('''
Список команд:
1. Добавить оценку ученика по предмету
2. Вывести средний балл по всем предметам по каждому ученику
3. Вывести все оценки по всем ученикам
4. Выход из программы
5. Удалить оценку ученика по предмету
6. Изменить (редактировать) оценку ученика
7. Добавить или удалить ученика
8. Добавить или удалить предмет
9. Вывести все оценки определённого ученика
10. Вывести средний балл по каждому предмету определённого ученика
''')

while True:
    try:
        command = int(input('Введите команду: '))
    except ValueError:
        print('Ошибка: нужно ввести число!')
        continue

    # --- 1. Добавить оценку ---
    if command == 1:
        student = input('Введите имя ученика: ')
        class_ = input('Введите предмет: ')
        try:
            mark = int(input('Введите оценку (1–5): '))
        except ValueError:
            print('Ошибка: нужно ввести число от 1 до 5.')
            continue

        if student in students_marks and class_ in students_marks[student]:
            students_marks[student][class_].append(mark)
            print(f'Для {student} по предмету {class_} добавлена оценка {mark}')
        else:
            print('Ошибка: неверное имя ученика или предмет.')

    # --- 2. Средний балл по всем предметам для всех ---
    elif command == 2:
        for student in students:
            print(f"\n{student}")
            for class_ in classes:
                marks = students_marks[student][class_]
                avg = round(sum(marks) / len(marks), 2)
                print(f"{class_}: {avg}")
        print()

    # --- 3. Все оценки всех ---
    elif command == 3:
        for student in students:
            print(f"\n{student}")
            for class_ in classes:
                print(f"\t{class_}: {students_marks[student][class_]}")
        print()

    # --- 4. Выход ---
    elif command == 4:
        print('Выход из программы...')
        break

    # --- 5. Удалить оценку ---
    elif command == 5:
        student = input('Введите имя ученика: ')
        class_ = input('Введите предмет: ')
        if student in students_marks and class_ in students_marks[student]:
            print(f"Оценки {student} по {class_}: {students_marks[student][class_]}")
            try:
                index = int(input('Введите номер оценки для удаления (начиная с 1): ')) - 1
                removed = students_marks[student][class_].pop(index)
                print(f'Удалена оценка {removed}')
            except (ValueError, IndexError):
                print('Ошибка: неверный номер оценки.')
        else:
            print('Ошибка: неверное имя ученика или предмет.')

    # --- 6. Изменить оценку ---
    elif command == 6:
        student = input('Введите имя ученика: ')
        class_ = input('Введите предмет: ')
        if student in students_marks and class_ in students_marks[student]:
            print(f"Оценки {student} по {class_}: {students_marks[student][class_]}")
            try:
                index = int(input('Введите номер оценки для изменения (начиная с 1): ')) - 1
                new_mark = int(input('Введите новую оценку: '))
                students_marks[student][class_][index] = new_mark
                print(f'Оценка обновлена: {students_marks[student][class_]}')
            except (ValueError, IndexError):
                print('Ошибка: неверный номер или значение оценки.')
        else:
            print('Ошибка: неверное имя ученика или предмет.')

    # --- 7. Добавить / удалить ученика ---
    elif command == 7:
        action = input('Введите действие (добавить/удалить): ').lower()
        if action == 'добавить':
            name = input('Введите имя нового ученика: ')
            if name not in students_marks:
                students.append(name)
                students.sort()
                students_marks[name] = {c: [] for c in classes}
                print(f'Ученик {name} добавлен.')
            else:
                print('Такой ученик уже есть.')
        elif action == 'удалить':
            name = input('Введите имя ученика для удаления: ')
            if name in students_marks:
                students.remove(name)
                del students_marks[name]
                print(f'Ученик {name} удалён.')
            else:
                print('Такого ученика нет.')

    # --- 8. Добавить / удалить предмет ---
    elif command == 8:
        action = input('Введите действие (добавить/удалить): ').lower()
        if action == 'добавить':
            new_class = input('Введите название нового предмета: ')
            if new_class not in classes:
                classes.append(new_class)
                for student in students_marks:
                    students_marks[student][new_class] = []
                print(f'Предмет {new_class} добавлен.')
            else:
                print('Такой предмет уже есть.')
        elif action == 'удалить':
            del_class = input('Введите название предмета для удаления: ')
            if del_class in classes:
                classes.remove(del_class)
                for student in students_marks:
                    del students_marks[student][del_class]
                print(f'Предмет {del_class} удалён.')
            else:
                print('Такого предмета нет.')

    # --- 9. Все оценки определённого ученика ---
    elif command == 9:
        student = input('Введите имя ученика: ')
        if student in students_marks:
            print(f'\nОценки ученика {student}:')
            for class_, marks in students_marks[student].items():
                print(f'{class_}: {marks}')
        else:
            print('Такого ученика нет.')

    # --- 10. Средний балл по предметам у ученика ---
    elif command == 10:
        student = input('Введите имя ученика: ')
        if student in students_marks:
            print(f'\nСредний балл ученика {student}:')
            for class_, marks in students_marks[student].items():
                if marks:
                    avg = round(sum(marks) / len(marks), 2)
                    print(f'{class_}: {avg}')
                else:
                    print(f'{class_}: нет оценок')
        else:
            print('Такого ученика нет.')

    else:
        print('Неизвестная команда. Попробуйте снова.')
```
